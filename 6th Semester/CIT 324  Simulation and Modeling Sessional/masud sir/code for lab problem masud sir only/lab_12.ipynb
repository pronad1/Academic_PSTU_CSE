# Lab 12: Random Variates Using the Inverse Transform Function

**Problem Statement:**
There are several general approaches to generating univariate Random Variables from the distribution function. Write a program which generates the desired number of random variates using the inverse transform function.


### Theory and Algorithm

The **Inverse Transform Method** is a fundamental way to generate random numbers from any probability distribution, given its Cumulative Distribution Function (CDF).

**Steps:**
1. Let the target distribution have a CDF denoted as $F(x) = P(X \le x)$.
2. Calculate the inverse of the CDF, denoted as $F^{-1}(u)$.
3. Generate a uniform random number $U \sim U(0,1)$.
4. Compute the random variate $X = F^{-1}(U)$.

**Example Chosen: Weibull Distribution**
- CDF: $F(x) = 1 - e^{-(x/\lambda)^k}$
- Inverse CDF: $F^{-1}(u) = \lambda (-\ln(1-u))^{1/k}$


```
import numpy as np
import matplotlib.pyplot as plt

def generate_weibull_inverse_transform(lam, k, num_samples):
    """
    Generates random variates for a Weibull distribution using Inverse Transform.
    lam: Scale parameter (lambda)
    k: Shape parameter
    """
    # Step 1 & 3: Generate Uniform Random Numbers U(0,1)
    u = np.random.uniform(0, 1, num_samples)
    
    # Step 2 & 4: Apply Inverse CDF
    # X = lambda * (-ln(1 - U))^(1/k)
    # Note: Since U is uniform(0,1), (1-U) is also uniform(0,1). 
    # We can just use U instead of 1-U for simplicity, but we'll stick to the exact formula.
    variates = lam * (-np.log(1 - u))**(1/k)
    
    return variates

# Parameters for Weibull
lam_param = 5.0  # Scale
k_param = 1.5    # Shape
n_samples = 5000

# Generate variates
weibull_variates = generate_weibull_inverse_transform(lam_param, k_param, n_samples)
print(f"Generated {n_samples} Weibull variates via Inverse Transform.")

# Plotting the empirical vs theoretical
plt.figure(figsize=(10, 6))
# Empirical
count, bins, ignored = plt.hist(weibull_variates, bins=50, density=True, color='#9467bd', edgecolor='black', alpha=0.6, label='Empirical Data (Inverse Transform)')

# Theoretical PDF for Weibull: f(x) = (k/lam) * (x/lam)^(k-1) * e^(-(x/lam)^k)
x_vals = np.linspace(0, max(bins), 200)
theoretical_pdf = (k_param / lam_param) * ((x_vals / lam_param)**(k_param - 1)) * np.exp(-(x_vals / lam_param)**k_param)

plt.plot(x_vals, theoretical_pdf, color='black', linewidth=2, linestyle='--', label='Theoretical Weibull PDF')
plt.title(f"Weibull Distribution via Inverse Transform (\u03bb={lam_param}, k={k_param})", fontsize=14, fontweight='bold')
plt.xlabel("Value (x)", fontsize=12)
plt.ylabel("Density f(x)", fontsize=12)
plt.legend()
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.show()
```
