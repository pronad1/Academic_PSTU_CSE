# Lab 13: Generate 5,000 Exponential Random Variates

**Problem Statement:**
Generate 5,000 exponentially distributed random numbers using the Inverse Transform technique. Plot a histogram of the resulting numbers and overlay the theoretical exponential probability density function.


### Theory and Algorithm (Exponential Specific)

As learned in Lab 12, the Inverse Transform Method requires the inverse CDF.

**Exponential Distribution:**
- PDF: $f(x) = \lambda e^{-\lambda x}$
- CDF: $F(x) = 1 - e^{-\lambda x}$

**Finding Inverse CDF $F^{-1}(U)$**:
1. Set $U = 1 - e^{-\lambda x}$
2. $e^{-\lambda x} = 1 - U$
3. $-\lambda x = \ln(1 - U)$
4. $x = -\\frac{1}{\lambda} \ln(1 - U)$

Since $U$ is a uniform random variable in $[0,1)$, $(1-U)$ is identically distributed to $U$. 
Thus, a slightly more optimized formula commonly used is:
$$X = -\\frac{1}{\lambda} \ln(U)$$


```
import numpy as np
import matplotlib.pyplot as plt

def generate_exponential_inverse_transform(lam, num_samples):
    """
    Generates random variates for an Exponential distribution using Inverse Transform.
    lam: Rate parameter (lambda)
    """
    # Generate Uniform Random Numbers U(0,1)
    # Using np.random.rand which is uniform in [0, 1). We add small epsilon to avoid ln(0).
    u = np.random.uniform(1e-10, 1, num_samples)
    
    # Apply Inverse CDF: X = (-1 / lambda) * ln(U)
    variates = (-1.0 / lam) * np.log(u)
    
    return variates

# Parameters
lambda_rate = 2.0  # Rate parameter
n_samples = 5000

# Generate 5,000 variates
expo_variates = generate_exponential_inverse_transform(lambda_rate, n_samples)
print(f"Generated {n_samples} Exponential variates via Inverse Transform.")

# Plotting
plt.figure(figsize=(10, 6))

# Histogram (Empirical)
plt.hist(expo_variates, bins=50, density=True, color='#ff7f0e', edgecolor='black', alpha=0.7, label='Empirical Histogram')

# Theoretical PDF
x_vals = np.linspace(0, max(expo_variates), 200)
theoretical_pdf = lambda_rate * np.exp(-lambda_rate * x_vals)

# Overlay PDF
plt.plot(x_vals, theoretical_pdf, color='blue', linewidth=2, linestyle='--', label=f'Theoretical PDF (\u03bb={lambda_rate})')

plt.title("5,000 Exponential Variates using Inverse Transform", fontsize=14, fontweight='bold')
plt.xlabel("Value (x)", fontsize=12)
plt.ylabel("Density f(x)", fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.4)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.show()
```
