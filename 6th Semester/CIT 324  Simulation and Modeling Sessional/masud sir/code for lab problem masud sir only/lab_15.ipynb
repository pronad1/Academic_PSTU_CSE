# Lab 15: Confidence Interval Estimation

**Problem Statement:**
Calculate a 95% confidence interval for the mean system response (e.g., average waiting time) by running multiple independent replications of a simulation model.


### Theory and Formulas

Because simulations involve random variables, the output (like average waiting time) is also a random variable. Running a simulation once only gives a **point estimate**. To understand the reliability of this estimate, we use **Confidence Intervals (CI)**.

**Method of Independent Replications**:
1. Run the simulation $n$ independent times (using different random seeds).
2. Collect the mean system response $X_i$ from each replication $i$.
3. Calculate the sample mean $\\bar{X}$ and sample standard deviation $S$ of these $n$ responses.
4. The $(1 - \\alpha) \\times 100\\%$ Confidence Interval is:
   $$\\bar{X} \\pm t_{n-1, 1-\\alpha/2} \\frac{S}{\\sqrt{n}}$$

Where $t$ is the critical value from the Student's t-distribution with $n-1$ degrees of freedom. For a 95% CI, $\\alpha = 0.05$.


```
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

np.random.seed(42)

# ==========================================
# 1. Define a Mock Simulation Model
# ==========================================
def run_simulation():
    """
    Mock simulation of a queue. 
    Returns the average waiting time for 1000 customers.
    (We'll just generate random waiting times from a known distribution to simulate this)
    """
    # Let's say true waiting time is Exponential with mean 15.0
    wait_times = np.random.exponential(15.0, size=1000)
    return np.mean(wait_times)

# ==========================================
# 2. Perform Independent Replications
# ==========================================
n_replications = 30
responses = []

for i in range(n_replications):
    avg_wait = run_simulation()
    responses.append(avg_wait)
    
# ==========================================
# 3. Calculate 95% Confidence Interval
# ==========================================
sample_mean = np.mean(responses)
sample_std = np.std(responses, ddof=1) # ddof=1 for sample standard deviation

alpha = 0.05
# t-critical value for two-tailed test
t_crit = stats.t.ppf(1 - alpha/2, df=n_replications - 1)

margin_of_error = t_crit * (sample_std / np.sqrt(n_replications))

ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print("--- Confidence Interval Estimation ---")
print(f"Number of Replications (n) : {n_replications}")
print(f"Sample Mean (Point Est)    : {sample_mean:.4f}")
print(f"Sample Standard Deviation  : {sample_std:.4f}")
print(f"t-critical value (95%)     : {t_crit:.4f}")
print(f"Margin of Error            : {margin_of_error:.4f}")
print("-" * 38)
print(f"95% Confidence Interval    : [{ci_lower:.4f} , {ci_upper:.4f}]")

# ==========================================
# 4. Visualization
# ==========================================
plt.figure(figsize=(8, 2))
plt.errorbar(sample_mean, 1, xerr=margin_of_error, fmt='o', color='#d62728', 
             capsize=8, capthick=2, markersize=8, label='95% CI')
plt.axvline(15.0, color='gray', linestyle='--', label='True Mean (15.0)')
plt.yticks([])
plt.title("95% Confidence Interval for Mean Waiting Time", fontweight='bold')
plt.xlabel("Average Waiting Time")
plt.legend()
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['left'].set_visible(False)
plt.show()
```
