# Lab 02: Inventory Management Simulation

**Problem Statement:**
Inventory management is a crucial aspect of any business that deals with physical goods. Write a program to simulate an inventory system.


### Theory and Approach (Periodic Review System)

We will simulate a **Periodic Review Inventory System (M, L)**:
- **M**: Maximum inventory level.
- **L**: Review period (in days).

Every $L$ days, the inventory is reviewed. If the inventory level $I$ is less than $M$, an order of size $Q = M - I$ is placed to bring it back up to $M$.
- **Demand**: Random daily demand (e.g., uniformly distributed).
- **Lead Time**: Time between placing an order and receiving it (assumed 0 for simplicity in this basic model, but easily extensible).
- **Costs**:
    - **Holding Cost**: Cost per unit of inventory held per day.
    - **Shortage Cost**: Cost per unit of unmet demand.

Our goal is to simulate this system for a number of days and calculate the total costs.


```
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Parameters
MAX_INVENTORY = 50
REVIEW_PERIOD = 5       # Review every 5 days
SIMULATION_DAYS = 100

HOLDING_COST_PER_UNIT = 0.5
SHORTAGE_COST_PER_UNIT = 2.0

# State variables
current_inventory = MAX_INVENTORY
total_holding_cost = 0.0
total_shortage_cost = 0.0

# Data for plotting
inventory_levels = []
days = list(range(1, SIMULATION_DAYS + 1))

print("--- Inventory Simulation Started ---")

for day in days:
    # 1. Start of day: check if it's a review day
    if day % REVIEW_PERIOD == 1 and day != 1:
        order_quantity = MAX_INVENTORY - current_inventory
        if order_quantity > 0:
            current_inventory += order_quantity
            # print(f"Day {day}: Ordered {order_quantity} units.")
            
    # 2. Demand occurs (Uniformly distributed between 0 and 15)
    daily_demand = np.random.randint(0, 16)
    
    # 3. Fulfill demand and calculate costs
    if current_inventory >= daily_demand:
        current_inventory -= daily_demand
        total_holding_cost += current_inventory * HOLDING_COST_PER_UNIT
    else:
        shortage = daily_demand - current_inventory
        total_shortage_cost += shortage * SHORTAGE_COST_PER_UNIT
        current_inventory = 0  # Assuming lost sales (no backlogging)
        
    inventory_levels.append(current_inventory)

total_cost = total_holding_cost + total_shortage_cost

print(f"Simulation completed for {SIMULATION_DAYS} days.")
print(f"Total Holding Cost  : ${total_holding_cost:.2f}")
print(f"Total Shortage Cost : ${total_shortage_cost:.2f}")
print(f"Total Cost          : ${total_cost:.2f}")
```

```
plt.figure(figsize=(12, 5))
plt.plot(days, inventory_levels, marker='o', linestyle='-', color='#2ca02c', markersize=4)
plt.axhline(MAX_INVENTORY, color='red', linestyle='--', label='Max Inventory (M)')
plt.title("Inventory Level Over Time (Periodic Review)", fontsize=14, fontweight='bold')
plt.xlabel("Day", fontsize=12)
plt.ylabel("Inventory Level", fontsize=12)
plt.legend()
plt.grid(alpha=0.5)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```
