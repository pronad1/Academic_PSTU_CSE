# Lab 05: Network Packet Delivery (Binomial Distribution)

**Problem Statement:**
Scenario - A network router transmits data packets in batches. Each individual packet has a success probability of p = 0.90 of reaching its destination. A single batch consists of n = 15 independent packets. 
Example Data: A simulation of 5 separate batches reveals the number of successfully delivered packets per batch: 14, 13, 15, 12, 14. 

**Tasks:**
1. Identify parameters: Define the values for n, p, and q (failure probability). 
2. Exact Probability: Calculate the probability that exactly 13 packets arrive safely in a batch. 
3. Expected Value & Variance: Compute the theoretical mean $\mu$ and variance $\sigma^2$ for the batch. 
4. Data Verification: Calculate the average number of successful packets from the example data and compare it to the theoretical mean.


### Theory 

Because we have a fixed number of independent Bernoulli trials ($n=15$) each with the same probability of success ($p=0.90$), the number of successful packets in a batch follows a **Binomial Distribution**: $X \sim B(n, p)$.

- **Probability Mass Function**: $P(X = k) = \binom{n}{k} p^k q^{n-k}$
- **Mean**: $\mu = np$
- **Variance**: $\sigma^2 = npq$


```
import math
import numpy as np
import scipy.stats as stats

# 1. Identify Parameters
n = 15
p = 0.90
q = 1 - p

print("1. Parameters:")
print(f"   n (batch size) = {n}")
print(f"   p (success prob) = {p}")
print(f"   q (failure prob) = {q:.2f}")

# 2. Exact Probability P(X = 13)
k = 13
# Using combination formula: n! / (k! * (n-k)!) * p^k * q^(n-k)
combinations = math.comb(n, k)
prob_13 = combinations * (p**k) * (q**(n-k))

# Alternatively using scipy:
# prob_13_scipy = stats.binom.pmf(k, n, p)

print("\n2. Exact Probability:")
print(f"   P(X = 13) = {prob_13:.4f}")

# 3. Expected Value and Variance
mu = n * p
variance = n * p * q

print("\n3. Theoretical Metrics:")
print(f"   Expected Value (μ) = {mu:.2f}")
print(f"   Variance (σ²)      = {variance:.2f}")

# 4. Data Verification
example_data = [14, 13, 15, 12, 14]
sample_mean = np.mean(example_data)

print("\n4. Data Verification:")
print(f"   Example Data           = {example_data}")
print(f"   Sample Mean            = {sample_mean:.2f}")
print(f"   Theoretical Mean       = {mu:.2f}")
print(f"   Difference             = {abs(sample_mean - mu):.2f}")
```
