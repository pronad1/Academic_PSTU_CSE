# Lab 06: Poisson Distribution

**Problem Statement:**
A customer care center receives 5 calls per hour. Compute the probability of attending zero calls, one call, two calls, \u2026.., ten calls per hour. Simulate the probability mass function respect to the number of receiving call. Show the probability mass function graphs considering 10 and 15 calls per hour.


### Theory and Formulas

The **Poisson Distribution** models the number of events occurring in a fixed interval of time or space if these events happen with a known constant mean rate and independently of the time since the last event.

- **Probability Mass Function (PMF)**: $P(X = k) = \\frac{\lambda^k e^{-\lambda}}{k!}$
- **Rate Parameter ($\lambda$)**: The expected number of events in the interval.

For our problem:
1. Base scenario: $\lambda = 5$ calls/hour. We need $P(X=k)$ for $k \in \{0, 1, \dots, 10\}$.
2. Additional scenarios: $\lambda = 10$ and $\lambda = 15$ calls/hour for graphing.


```
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# 1. Compute probabilities for lambda = 5
lambda_5 = 5
k_values = np.arange(0, 11) # 0 to 10
pmf_5 = stats.poisson.pmf(k_values, lambda_5)

print(f"--- Poisson Probabilities (λ = {lambda_5} calls/hr) ---")
for k, p in zip(k_values, pmf_5):
    print(f"P(X = {k:2d}) = {p:.4f}")
    
# 2. Simulate and Graph PMF for lambda = 10 and 15
lambdas = [10, 15]
x = np.arange(0, 30)

plt.figure(figsize=(12, 6))

for lam in lambdas:
    pmf_values = stats.poisson.pmf(x, lam)
    
    # We use stem plot or bar plot for discrete distributions (PMF)
    # Adding a slight offset to x for better visibility when overlapping
    offset = -0.2 if lam == 10 else 0.2
    plt.bar(x + offset, pmf_values, width=0.4, label=f'$\lambda = {lam}$', alpha=0.8)

plt.title("Poisson Probability Mass Function (PMF)", fontsize=14, fontweight='bold')
plt.xlabel("Number of Calls per Hour (k)", fontsize=12)
plt.ylabel("Probability P(X = k)", fontsize=12)
plt.xticks(x)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```
