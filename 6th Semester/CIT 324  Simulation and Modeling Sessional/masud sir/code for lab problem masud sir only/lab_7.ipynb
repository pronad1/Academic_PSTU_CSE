# Lab 07: Normal Distribution

**Problem Statement:**
1. Show unimodal and multimodal density curves of normal distribution. 
2. Generate random sample with sample size 200, which follows a normal distribution with mean 100 and standard deviation 20. 
3. The distribution of diastolic blood pressure for men is normally distributed with a mean of about 80 and a standard deviation of 20. 
4. A histogram of the distribution of blood pressures for all mean displays a normal distribution with bell shape.


### Theory

The **Normal (Gaussian) Distribution** is a continuous probability distribution characterized by its mean ($\mu$) and standard deviation ($\sigma$).
- **PDF**: $f(x) = \\frac{1}{\sigma \sqrt{2\pi}} e^{-\\frac{1}{2}(\\frac{x-\mu}{\sigma})^2}$

A standard normal distribution is **unimodal** (one peak). A **multimodal** distribution (e.g., bimodal) can be created by a mixture of two normal distributions with different means.


```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.linspace(20, 180, 1000)
unimodal_pdf = stats.norm.pdf(x, loc=100, scale=20)
bimodal_pdf = 0.5 * stats.norm.pdf(x, loc=70, scale=10) + 0.5 * stats.norm.pdf(x, loc=130, scale=15)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(x, unimodal_pdf, color='blue', linewidth=2)
axes[0].fill_between(x, unimodal_pdf, alpha=0.3, color='blue')
axes[0].set_title("Unimodal Normal Distribution")

axes[1].plot(x, bimodal_pdf, color='purple', linewidth=2)
axes[1].fill_between(x, bimodal_pdf, alpha=0.3, color='purple')
axes[1].set_title("Bimodal (Multimodal) Distribution")

for ax in axes:
    ax.set_ylabel("Density")
    ax.grid(alpha=0.4)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
plt.tight_layout()
plt.show()
sample_200 = np.random.normal(loc=100, scale=20, size=200)


blood_pressures = np.random.normal(loc=80, scale=20, size=5000)

plt.figure(figsize=(10, 6))
count, bins, ignored = plt.hist(blood_pressures, bins=40, density=True, alpha=0.6, color='#2ca02c', edgecolor='black', label='Sampled Data Histogram')
x_bell = np.linspace(20, 140, 100)
bell_curve = stats.norm.pdf(x_bell, loc=80, scale=20)
plt.plot(x_bell, bell_curve, color='red', linewidth=2, label='Theoretical PDF $N(80, 20^2)$')

plt.title("Diastolic Blood Pressure Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Blood Pressure", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.legend()
plt.grid(alpha=0.5)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.show()
```
