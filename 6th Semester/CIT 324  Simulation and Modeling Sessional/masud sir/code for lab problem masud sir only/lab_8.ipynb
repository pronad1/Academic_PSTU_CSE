# Lab 08: Exponential Distribution

**Problem Statement:**
There were few waves in COVID-19 pandemic. Let a wave occur every 100 days in Bangladesh, on average. After a wave occurs, find the probability using Exponential distribution that it will take more than 120 days for the next wave to occur. Simulate several Exponential distribution using rate parameters 0.5, 1.0, 2.0, and 4.0.


### Theory and Formulas

The **Exponential Distribution** is often used to model the time between independent events that happen at a constant average rate.

- **Rate Parameter ($\lambda$)**: The average number of events per unit time. 
- **Mean (Expected Time)**: $\mu = \frac{1}{\lambda}$
- **Probability Density Function (PDF)**: $f(x) = \lambda e^{-\lambda x}$ for $x \ge 0$
- **Cumulative Distribution Function (CDF)**: $P(X \le x) = 1 - e^{-\lambda x}$
- **Survival Function (Tail Probability)**: $P(X > x) = 1 - P(X \le x) = e^{-\lambda x}$

**For the COVID-19 wave problem:**
- Mean time between waves: $\mu = 100$ days.
- Rate $\lambda = \frac{1}{100} = 0.01$ waves/day.
- We need to find $P(X > 120)$.


```
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# ==========================================
# 1. COVID-19 Wave Probability
# ==========================================
mean_days = 100
lambda_rate = 1 / mean_days
x_target = 120

# We want P(X > 120)
# P(X > x) = e^(-lambda * x)
prob_more_than_120 = np.exp(-lambda_rate * x_target)

# Using scipy
# prob_more_than_120_scipy = 1 - stats.expon.cdf(x_target, scale=mean_days)

print("--- COVID-19 Wave Probability ---")
print(f"Mean time between waves (μ) : {mean_days} days")
print(f"Rate parameter (λ)          : {lambda_rate:.4f} waves/day")
print(f"Probability P(X > 120 days)  : {prob_more_than_120:.4f} ({prob_more_than_120*100:.2f}%)")
print("-" * 33)

# ==========================================
# 2. Simulate Several Exponential Distributions
# ==========================================
rate_parameters = [0.5, 1.0, 2.0, 4.0]
x_vals = np.linspace(0, 5, 1000)

plt.figure(figsize=(10, 6))

for lam in rate_parameters:
    # Note: scipy.stats.expon uses 'scale' which is 1/lambda
    scale_param = 1 / lam
    pdf_vals = stats.expon.pdf(x_vals, scale=scale_param)
    
    plt.plot(x_vals, pdf_vals, linewidth=2, label=f'$\lambda$ = {lam} ($\mu$ = {scale_param:.2f})')

plt.title("Exponential Distribution PDFs for Various Rates (λ)", fontsize=14, fontweight='bold')
plt.xlabel("x", fontsize=12)
plt.ylabel("Probability Density f(x)", fontsize=12)
plt.xlim(0, 5)
plt.ylim(0, 4.5)
plt.legend()
plt.grid(alpha=0.4)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.show()
```
