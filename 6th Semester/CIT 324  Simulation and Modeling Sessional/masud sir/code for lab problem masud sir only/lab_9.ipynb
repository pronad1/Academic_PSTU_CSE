# Lab 09: Empirical Input Modeling

**Problem Statement:**
Given a raw dataset of customer arrival times at a call center, fit the data to theoretical distributions (e.g., Normal, Weibull, or Lognormal) using **Maximum Likelihood Estimation (MLE)**. Validate the fit with **Q-Q plots**.


### Theory

**Input Modeling** is the process of finding a probability distribution that best represents empirical (observed) data.
1. **Maximum Likelihood Estimation (MLE)**: A method to estimate the parameters of a statistical model. It maximizes the likelihood function, making the observed data most probable.
2. **Q-Q Plot (Quantile-Quantile Plot)**: A graphical tool to compare two probability distributions. If the data perfectly follows the theoretical distribution, the points will lie exactly on the $y = x$ reference line.


```
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

np.random.seed(42)

# ==========================================
# 1. Generate Synthetic 'Raw' Data
# ==========================================
# We'll generate data that is actually Weibull distributed to see which fit is best.
true_shape = 1.5
true_scale = 10.0
raw_data = stats.weibull_min.rvs(true_shape, scale=true_scale, size=500)

print(f"Generated {len(raw_data)} data points for arrival times.\n")

# ==========================================
# 2. Fit Distributions using MLE
# ==========================================
# Normal Fit
mu_norm, std_norm = stats.norm.fit(raw_data)
print(f"Normal MLE Fit   : μ = {mu_norm:.2f}, σ = {std_norm:.2f}")

# Lognormal Fit
shape_lognorm, loc_lognorm, scale_lognorm = stats.lognorm.fit(raw_data, floc=0)
print(f"Lognormal MLE Fit: shape = {shape_lognorm:.2f}, scale = {scale_lognorm:.2f}")

# Weibull Fit
shape_weibull, loc_weibull, scale_weibull = stats.weibull_min.fit(raw_data, floc=0)
print(f"Weibull MLE Fit  : shape = {shape_weibull:.2f}, scale = {scale_weibull:.2f}")

# ==========================================
# 3. Validation via Q-Q Plots
# ==========================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Normal Q-Q Plot
stats.probplot(raw_data, dist="norm", sparams=(mu_norm, std_norm), plot=axes[0])
axes[0].set_title("Q-Q Plot: Normal Fit")

# Lognormal Q-Q Plot
stats.probplot(raw_data, dist="lognorm", sparams=(shape_lognorm,), plot=axes[1])
axes[1].set_title("Q-Q Plot: Lognormal Fit")

# Weibull Q-Q Plot
stats.probplot(raw_data, dist="weibull_min", sparams=(shape_weibull,), plot=axes[2])
axes[2].set_title("Q-Q Plot: Weibull Fit")

for ax in axes:
    ax.grid(alpha=0.4)

plt.tight_layout()
plt.show()

# ==========================================
# 4. Histogram Overlays
# ==========================================
x = np.linspace(0, max(raw_data), 1000)

plt.figure(figsize=(10, 6))
plt.hist(raw_data, bins=30, density=True, alpha=0.5, color='gray', label='Empirical Data')

plt.plot(x, stats.norm.pdf(x, mu_norm, std_norm), label='Normal Fit', linestyle='--')
plt.plot(x, stats.lognorm.pdf(x, shape_lognorm, loc_lognorm, scale_lognorm), label='Lognormal Fit', linestyle='--')
plt.plot(x, stats.weibull_min.pdf(x, shape_weibull, loc_weibull, scale_weibull), label='Weibull Fit', linewidth=2, color='red')

plt.title("Distribution Fits on Empirical Data", fontsize=14, fontweight='bold')
plt.xlabel("Arrival Time", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.legend()
plt.grid(alpha=0.4)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.show()
```
