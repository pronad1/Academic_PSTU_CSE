```python
import random
import matplotlib.pyplot as plt

# --- CONFIGURATION ---
num_steps = 100  # Number of steps to take

# Lists to store the path (Start at 0,0)
x_path = [0]
y_path = [0]

# --- SIMULATION ---
for _ in range(num_steps):
    # Get current position (last item in the list)
    current_x = x_path[-1]
    current_y = y_path[-1]
    
    # Randomly choose a direction (0, 1, 2, 3)
    # 0 = Up (y+1), 1 = Down (y-1), 2 = Right (x+1), 3 = Left (x-1)
    direction = random.randint(0, 3)
    
    if direction == 0:   # Up
        y_path.append(current_y + 1)
        x_path.append(current_x)
    elif direction == 1: # Down
        y_path.append(current_y - 1)
        x_path.append(current_x)
    elif direction == 2: # Right
        x_path.append(current_x + 1)
        y_path.append(current_y)
    elif direction == 3: # Left
        x_path.append(current_x - 1)
        y_path.append(current_y)

# --- PLOTTING ---
plt.figure(figsize=(8, 8))

# Plot the entire path
plt.plot(x_path, y_path, color='blue', linewidth=0.8, alpha=0.7, label='Path')

# Mark the START point (Green)
plt.plot(0, 0, 'go', markersize=10, label='Start (0,0)')

# Mark the END point (Red)
plt.plot(x_path[-1], y_path[-1], 'ro', markersize=10, label='End')

# Formatting
plt.title(f"2D Random Walk ({num_steps} Steps)")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.axis('equal') # Keeps the aspect ratio square so the path isn't distorted

plt.show()

# --- STATISTICS ---
# Calculate Euclidean distance from start
final_dist = (x_path[-1]**2 + y_path[-1]**2)**0.5
print(f"Total Steps: {num_steps}")
print(f"Final Position: ({x_path[-1]}, {y_path[-1]})")
print(f"Distance from Start: {final_dist:.2f}")
```
