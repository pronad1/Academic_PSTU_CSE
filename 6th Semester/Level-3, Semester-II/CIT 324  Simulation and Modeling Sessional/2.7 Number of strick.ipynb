```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
length = 1000
height = 500
num_points = 1000

x_min, x_max = -length/2, length/2
y_min, y_max = -height/2, height/2
```

```python
gen_x = np.random.uniform(x_min * 1.5, x_max * 1.5, num_points)
gen_y = np.random.uniform(y_min * 1.5, y_max * 1.5, num_points)

inside_indices = (gen_x >= x_min) & (gen_x <= x_max) & (gen_y >= y_min) & (gen_y <= y_max)

x_in, y_in = gen_x[inside_indices], gen_y[inside_indices]
x_out, y_out = gen_x[~inside_indices], gen_y[~inside_indices]

hit_count = len(x_in)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(x_in, y_in, color="green", label="Hit", s=15)
plt.scatter(x_out, y_out, color="blue", label="Miss", s=15)

box_x = [x_min, x_max, x_max, x_min, x_min]
box_y = [y_min, y_min, y_max, y_max, y_min]
plt.plot(box_x, box_y, color="red", linewidth=2)

plt.legend()
plt.axis('equal')
plt.show()

print(hit_count, num_points)
print("Accuracy : ", (hit_count / num_points) * 100, "%")
```
