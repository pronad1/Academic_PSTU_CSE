```python
# Import lib
import random
import matplotlib.pyplot as plt
from shapely.geometry import Point, Polygon
```

```python
x_point = []
y_point = []

coords = [
    (25, 25),   # Bottom Left corner
    (40, 25),   # Start of bottom notch
    (40, 35),   # Notch goes up
    (55, 35),   # Notch goes right
    (55, 25),   # Notch goes down
    (75, 25),   # Bottom Right corner
    (75, 55),   # Start of right-side curve
    (85, 65),   # Curve bulge out
    (85, 75),   # Curve continues
    (75, 85),   # Curve comes back in
    (50, 95),   # Top "Head" peak
    (35, 85),   # Top Left side
    (35, 70),   # The "Step" down
    (25, 70),   # The "Step" left
    (25, 60),   # Down to the triangle cut
    (40, 50),   # The Triangle Cut (Inward point)
    (25, 40),   # The Triangle Cut (Return)
    (25, 25)    # Close the loop
]

map = Polygon(coords)

for i in coords:
    a = i[0]
    b = i[1]
    x_point.append(a)
    y_point.append(b)
x_point.append(x_point[0])
y_point.append(y_point[0])
plt.plot(x_point,y_point,color="red")
```

```python
x_point = []
y_point = []

coords = [
    (25, 25),   # Bottom Left corner
    (40, 25),   # Start of bottom notch
    (40, 35),   # Notch goes up
    (55, 35),   # Notch goes right
    (55, 25),   # Notch goes down
    (75, 25),   # Bottom Right corner
    (75, 55),   # Start of right-side curve
    (85, 65),   # Curve bulge out
    (85, 75),   # Curve continues
    (75, 85),   # Curve comes back in
    (50, 95),   # Top "Head" peak
    (35, 85),   # Top Left side
    (35, 70),   # The "Step" down
    (25, 70),   # The "Step" left
    (25, 60),   # Down to the triangle cut
    (40, 50),   # The Triangle Cut (Inward point)
    (25, 40),   # The Triangle Cut (Return)
    (25, 25)    # Close the loop
]

map = Polygon(coords)

for i in coords:
    a = i[0]
    b = i[1]
    x_point.append(a)
    y_point.append(b)
x_point.append(x_point[0])
y_point.append(y_point[0])
plt.plot(x_point,y_point,color="red")
n=0 
m=0

for i in range(1000):
    x=random.randint(25,90)
    y=random.randint(25,100)
    point=Point(x,y)
    n+=1
    if(map.contains(point)):
        m+=1
        plt.scatter(x,y,color="green")
    else:
        plt.scatter(x,y,color="blue")

plt.show()
print(m)
```
