```python
# Import lib
import random as rd
import numpy as np
```

```python
total_cost = 0
total_win = 0

for i in range(50):
    x=0
    h=0
    t=0
    while(1):

        x+=1
        total_cost += 1
        raises = rd.randint(0,1)
        if raises == 0:
            h+=1
        else:   
            t+=1
        dif=abs(h-t)
        if dif>=3:
            print(f'Won 8$ after cost of {x}$')
            total_win += 8
            break

print(f'Total cost: {total_cost}$')
print(f'Total win: {total_win}$')
```
