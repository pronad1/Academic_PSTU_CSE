```python
# Import lib
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3 - 2*x + 1
```

```python
n=0
m=0

for i in range(100):
    rx = random.randint(200,500)
    rx = rx/100
    ry = random.randint(0,150)

    n += 1
    if ry <= f(rx):
        m += 1
        plt.scatter(rx, ry, color='green')
    else:
        plt.scatter(rx, ry, color='blue')

x = []
y = []
i=0
h=0.2
while i<10:
    x.append(i)
    y.append(f(i))
    i += h
    i= round(i,2)

plt.plot(x, y, color='red')
plt.show()
```
