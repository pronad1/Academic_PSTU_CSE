```python
import random
import math
import matplotlib.pyplot as plt
```

```python
tot = 2000

inside_x = []
inside_y = []
outside_x = []
outside_y = []

points_inside_circle = 0

# --- SIMULATION ---
for _ in range(tot):
    x = random.random()
    y = random.random()
    
    if x**2 + y**2 <= 1:
        points_inside_circle += 1
        inside_x.append(x)
        inside_y.append(y)
    else:
        outside_x.append(x)
        outside_y.append(y)

pi_estimate = 4 * (points_inside_circle / tot)
error = abs(math.pi - pi_estimate)
print(f"Total Points: {tot}")
print(f"Points Inside: {points_inside_circle}")
print(f"Estimated Pi: {pi_estimate:.5f}")
print(f"Actual Pi:    {math.pi:.5f}")
print(f"Error:        {error:.5f}")
```

```python
# --- PLOTTING ---
plt.figure(figsize=(6, 6))

# Plot points
plt.scatter(inside_x, inside_y, color='green', s=1, label='Inside Circle')
plt.scatter(outside_x, outside_y, color='red', s=1, label='Outside Circle')

# Draw the actual circle boundary for reference
circle = plt.Circle((0, 0), 1, color='blue', fill=False, linewidth=2)
plt.gca().add_patch(circle)

# Formatting
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.gca().set_aspect('equal', adjustable='box')
plt.title(f"Monte Carlo Pi Estimation (N={tot})\nEstimated Pi = {pi_estimate:.5f}")
plt.legend(loc='upper right')
plt.show()
```
