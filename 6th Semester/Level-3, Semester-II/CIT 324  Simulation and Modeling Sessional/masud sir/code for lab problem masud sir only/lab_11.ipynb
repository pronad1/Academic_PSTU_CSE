# Lab 11: Linear Congruential Generator (LCG)

**Problem Statement:**
LCG is considered one of the basic yet best methods to generate Pseudo-Random Numbers. Write a program which generates the random number using the LCG.


### Theory and Formula

A **Linear Congruential Generator (LCG)** yields a sequence of pseudo-randomized numbers calculated with a discontinuous piecewise linear equation.

The generator is defined by the recurrence relation:
$$X_{n+1} = (a X_n + c) \\pmod m$$

Where:
- $X$ is the sequence of pseudo-random values.
- $m$: the modulus ($m > 0$)
- $a$: the multiplier ($0 < a < m$)
- $c$: the increment ($0 \\le c < m$)
- $X_0$: the seed or start value ($0 \\le X_0 < m$)

To get numbers distributed between 0 and 1 (Standard Uniform Distribution $U(0,1)$), we divide the generated sequence by $m$:
$$U_n = \\frac{X_n}{m}$$


```
import numpy as np
import matplotlib.pyplot as plt

def linear_congruential_generator(seed, a, c, m, num_samples):
    """
    Generates pseudo-random numbers using the LCG algorithm.
    """
    random_numbers = []
    x = seed
    for _ in range(num_samples):
        # The core LCG formula
        x = (a * x + c) % m
        
        # Normalize to [0, 1)
        u = x / m
        random_numbers.append(u)
        
    return random_numbers

# ==========================================
# 1. Generate Numbers
# ==========================================
# Standard parameters (e.g., similar to early C rand() implementations)
seed_value = 42
multiplier_a = 1103515245
increment_c = 12345
modulus_m = 2**31

num_to_generate = 5000

# Generate 5000 random numbers
pseudo_random_seq = linear_congruential_generator(seed_value, multiplier_a, increment_c, modulus_m, num_to_generate)

print(f"Generated {num_to_generate} pseudo-random numbers using LCG.")
print(f"First 5 numbers: {pseudo_random_seq[:5]}")

# ==========================================
# 2. Visualize Distribution
# ==========================================
plt.figure(figsize=(10, 6))
# Histogram should look perfectly flat (Uniform)
plt.hist(pseudo_random_seq, bins=50, density=True, color='#1f77b4', edgecolor='black', alpha=0.7)

# Theoretical uniform PDF line
plt.axhline(1.0, color='red', linestyle='--', linewidth=2, label='Theoretical U(0,1) PDF')

plt.title("Histogram of LCG Generated Numbers", fontsize=14, fontweight='bold')
plt.xlabel("Value", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.legend()
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.show()
```
