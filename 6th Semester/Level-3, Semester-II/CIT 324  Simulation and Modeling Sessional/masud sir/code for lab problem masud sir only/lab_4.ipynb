# Lab 04: Bernoulli Data Distribution

**Problem Statement:**
A Manufacturing and Reliability Lab is testing a newly produced batch of microchips. Each chip undergoes a strict operational test and either works perfectly (Success = 1) or fails (Failure = 0). Historical data shows that the probability of any given chip passing this test is p = 0.85. 

**Tasks:**
1. State the values of X and what they represent.
2. Calculate the probability of a defect, P(X=0).
3. Calculate the expected value, E[X].
4. Calculate the variance, Var(X).


### Theory and Answers

Let $X$ be the random variable representing the outcome of the operational test for a single microchip.

1. **Values of X**: 
   - $X = 1$ (Success: the chip works perfectly)
   - $X = 0$ (Failure: the chip is a defect)

2. **Probability of a defect $P(X=0)$**:
   Since the probability of success is $p = 0.85$, the probability of failure (defect) is $q = 1 - p$.
   $$P(X=0) = 1 - 0.85 = 0.15$$

3. **Expected Value $E[X]$**:
   For a Bernoulli distribution, the mean or expected value is simply $p$.
   $$E[X] = p = 0.85$$

4. **Variance $Var(X)$**:
   For a Bernoulli distribution, the variance is $p \times q$.
   $$Var(X) = p(1-p) = 0.85 \times 0.15 = 0.1275$$


```
# Python Implementation to calculate and verify the results
p = 0.85
q = 1 - p

print("--- Bernoulli Distribution Metrics ---")
print(f"Probability of Success (p) : {p}")
print(f"Probability of Defect  (q) : {q:.2f}  [ P(X=0) ]")
print("-" * 35)

expected_value = p
variance = p * q

print(f"Expected Value E[X]        : {expected_value:.2f}")
print(f"Variance Var(X)            : {variance:.4f}")
```
